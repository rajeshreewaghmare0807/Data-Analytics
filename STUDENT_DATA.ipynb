```python
import pandas as pd
import numpy as np
```

```python
old = pd.read_csv("STUDENT DATA.csv")
new = pd.read_csv("STUDENT DATA.csv")
```

```python

new.drop("S NO." , axis = 1 , inplace = True)
new
```

```python
new.drop([5 ],inplace=True)
new
```

```python
new.replace("NONE", np.nan, inplace=True)
new.fillna("Not Available", inplace=True)

new
```

```python
new["AGE"] = pd.to_numeric(new["AGE"], errors="coerce")
new["MARKS"] = pd.to_numeric(new["MARKS"], errors="coerce")
new
```

```python
new['AGE'].fillna(new['AGE'].mean(), inplace=True)
new['MARKS'].fillna(new['MARKS'].mean(), inplace=True)
new
```

```python
new["ROLL NO."] = pd.to_numeric(new["ROLL NO."], errors="coerce")

new["ROLL NO."] = new["ROLL NO."].interpolate()

new["ROLL NO."] = new["ROLL NO."].astype(int)

new
```

```python
# NAME fill
new.loc[new["ROLL NO."] == 103, "NAME"] = "SWATI"
new.loc[new["ROLL NO."] == 110, "NAME"] = "VAISHU"

# EMAIL ID correct
new.loc[new["ROLL NO."] == 108, "EMAIL ID"] = "nikku@gmail.com"
new.loc[new["ROLL NO."] == 114, "EMAIL ID"] = "yash@gmail.com"

# SUBJECT fill
new.loc[new["ROLL NO."] == 112, "SUBJECT"] = "Not Assigned"

# Final result
new
```

```python
new['AGE'] = new['AGE'].astype(int)
new['MARKS'] = new['MARKS'].astype(int)
new
```

```python
import matplotlib.pyplot as plt

x = new["SUBJECT"]
y = new["MARKS"]
L = new["NAME"]

c = [
    "red", "yellow", "orange", "blue", "purple", "black", "green",
    "white", "pink", "brown", "grey", "cyan", "magenta", "violet"
]

# Graph size
plt.figure(figsize=(15, 8))

plt.bar(
    x, y,
    label=L,
    color=c,
    edgecolor="black",
    width=0.5,
    linewidth=2,
    align="center",
    alpha=0.5
)

plt.legend()
plt.xlabel("SUBJECT")
plt.ylabel("MARKS")
plt.title("SUBJECT VS MARKS")

plt.show()
```
